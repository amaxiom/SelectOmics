# SelectOmics Test Suite

Runs the package test suite from a notebook.

The tests themselves live in `tests/*.py` and are the single source of truth.
This notebook used to inline a copy of them, which drifted out of step with the
package (it still asserted on `step4_target_retention` after Step 4 was
removed, so it failed on the first cell). It now just invokes pytest, so there
is only one copy to keep correct.

From a terminal the equivalent is:

```
pytest tests -v
```


````python
import subprocess
import sys
from pathlib import Path

# Run from the package root, whichever directory the notebook was opened in.
_here = Path.cwd()
ROOT = next((p for p in (_here, *_here.parents) if (p / "tests").is_dir()), _here)

print(f"Package root: {ROOT}")
result = subprocess.run(
    [sys.executable, "-m", "pytest", "tests", "-v", "--tb=short"],
    cwd=ROOT, capture_output=True, text=True,
)
print(result.stdout[-20000:])
if result.stderr.strip():
    print("--- stderr ---")
    print(result.stderr[-4000:])
print(f"\nexit code: {result.returncode}")
````

````python
# The integration tests are deselected by default (they run full pipelines).
# This runs them as well; expect a few minutes.
result = subprocess.run(
    [sys.executable, "-m", "pytest", "tests", "-v", "--tb=short", "-m", "integration"],
    cwd=ROOT, capture_output=True, text=True,
)
print(result.stdout[-20000:])
print(f"\nexit code: {result.returncode}")
````
